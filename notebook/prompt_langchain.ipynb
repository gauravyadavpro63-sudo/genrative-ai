dynamic prompts

```python

from dotenv import load_dotenv
load_dotenv()
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
llm=ChatGoogleGenerativeAI(
     model="gemini-3.5-flash"   
)
prompts=ChatPromptTemplate.from_messages([  ##class method
    {"role":"system","content":"You are translator and translate the input to {language}"},
    {"role":"user","content":"{query}"}
])
final_prompts=prompts.format_messages(language="hindi",query="i love javascript")

res=llm.invoke(final_prompts)
print(res.content[0]["text"]);
```

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
llm=ChatGoogleGenerativeAI(
     model="gemini-3.5-flash"   
)
prompts=ChatPromptTemplate.from_messages([  ##class method
    {"role":"system","content":"You are translator and translate the input to {language}"},
    {"role":"user","content":"{query}"}
])
res=llm.invoke(prompts.invoke({"language":"hindi","query":"i love u"}))
print(res.content[0]["text"])
```

chaining using pipe

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm=ChatGoogleGenerativeAI(
     model="gemini-3.5-flash"   
)
out=StrOutputParser()
def tranform_case(result):
    return result.upper()
prompts=ChatPromptTemplate.from_messages([  ##class method
    {"role":"system","content":"You are translator and translate the input to {language}"},
    {"role":"user","content":"{query}"}
])
chains=prompts|llm|out|tranform_case 
res=chains.invoke({"language":"hinglish","query":"i love langchain"})
print(res)
```
