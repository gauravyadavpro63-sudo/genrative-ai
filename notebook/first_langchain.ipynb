```python
from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(
    model="gemini-3.5-flash"
)
res=llm.invoke("Who is pm of india?")
print(res.content[0]["text"])  #it return object stored in res now we can get the attribute content and content contain list of dictionary 
print(res.content)
```

```python

from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(
    model="gemini-3.5-flash"
)
prompts=[
    ("system","You are python deveoper"),
    ("user","how to sort the array")
]
res=llm.invoke(prompts)
print(res.content)
```
